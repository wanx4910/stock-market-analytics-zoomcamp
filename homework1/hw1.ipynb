```python
import pandas as pd
import yfinance as yf
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
table = soup.find('table', {'id': 'constituents'})

df = pd.read_html(str(table))[0]
# Parse the 'Date first added' column
df['Date first added'] = pd.to_datetime(df['Date first added'], errors='coerce')
df['Year'] = df['Date first added'].dt.year

# Filter post-2020 and count additions per year
post_2020 = df[df['Year'] >= 2020]
addition_counts = post_2020['Year'].value_counts()
print(addition_counts)
# Answer corresponds to 2025 [1] (https://courses.datatalks.club/sma-zoomcamp-2026/homework/hw01)
```
